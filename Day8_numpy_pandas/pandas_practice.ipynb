```python
#importing libraries
import pandas as pd
import numpy as np
```

```python
# Create Dataframe
print("\n--- DATAFRAME ---")

data = {
    "employee_id": [101, 102, 103, 104, 105, 106, 107, 108],
    "name": [
        "Aakanksha",
        "Rahul",
        "Priya",
        "Sneha",
        "Amit",
        "Neha",
        "Rohan",
        "Pooja"
    ],
    "age": [24, 28, 26, 30, 25, 29, 27, 31],
    "department": [
        "IT",
        "HR",
        "IT",
        "Finance",
        "HR",
        "IT",
        "Finance",
        "Sales"
    ],
    "salary": [
        55000,
        65000,
        70000,
        75000,
        50000,
        80000,
        60000,
        90000
    ],
    "city": [
        "Pune",
        "Mumbai",
        "Pune",
        "Kolhapur",
        "Mumbai",
        "Pune",
        "Kolhapur",
        "Mumbai"
    ],
    "joining_date": [
        "2022-06-15",
        "2021-03-20",
        "2023-01-10",
        "2020-08-25",
        "2022-11-05",
        "2019-05-12",
        "2021-09-18",
        "2018-02-10"
    ],
    "experience": [2, 4, 1, 6, 3, 7, 4, 8]
}

df = pd.DataFrame(data)

print(df)
```

```python
#  READING AND WRITING DATA
print("\n--- READING AND WRITING DATA ---")

df.to_csv("employees.csv", index=False)

df = pd.read_csv("employees.csv")

print(df)

df.to_csv("employees_copy.csv", index=False)
```

```python
print("\n--- DATA INSPECTION ---")

print("\nHead:")
print(df.head())

print("\nTail:")
print(df.tail())

print("\nSample:")
print(df.sample(3))

print("\nShape:")
print(df.shape)

print("\nColumns:")
print(df.columns)

print("\nIndex:")
print(df.index)

print("\nData Types:")
print(df.dtypes)

print("\nInfo:")
df.info()

print("\nDescription:")
print(df.describe())

print("\nDepartment counts:")
print(df["department"].value_counts())

print("\nUnique departments:")
print(df["department"].nunique())

print("\nUnique cities:")
print(df["city"].nunique())
```

```python
print("\n--- 5. AXIS ---")

print("Column-wise sum:")
print(df[["age", "salary", "experience"]].sum(axis=0))

print("\nRow-wise sum:")
print(df[["age", "salary", "experience"]].sum(axis=1))
```

```python
print("\n--- SELECTING DATA ---")

print("Salary column:")
print(df["salary"])

print("\nName and salary:")
print(df[["name", "salary"]])

print("\nFirst 5 rows using iloc:")
print(df.iloc[:5])

print("\nFirst 3 rows and selected columns:")
print(df.loc[:2, ["name", "department", "salary"]])

print("\nSpecific row:")
print(df.iloc[2])
```

```python
print("\n--- FILTERING ---")

print("Salary greater than 60000:")
print(df[df["salary"] > 60000])

print("\nAge greater than 25:")
print(df[df["age"] > 25])

print("\nEmployees from Pune:")
print(df[df["city"] == "Pune"])

print("\nPune or Mumbai:")
print(df[df["city"].isin(["Pune", "Mumbai"])])

print("\nSalary between 50000 and 80000:")
print(df[df["salary"].between(50000, 80000)])

print("\nIT employees with salary > 60000:")
print(df[(df["department"] == "IT") & (df["salary"] > 60000)])

print("\nUsing query:")
print(df.query("salary > 60000 and age > 25"))
```

```python
print("\n--- MISSING VALUES ---")

missing_df = df.copy()

missing_df.loc[2, "salary"] = np.nan
missing_df.loc[4, "city"] = np.nan

print("Missing values:")
print(missing_df.isnull().sum())

print("\nRows containing missing values:")
print(missing_df[missing_df.isna().any(axis=1)])

print("\nFill salary with mean:")
missing_df["salary"] = missing_df["salary"].fillna(
    missing_df["salary"].mean()
)

print(missing_df)

print("\nFill city with mode:")
missing_df["city"] = missing_df["city"].fillna(
    missing_df["city"].mode()[0]
)

print(missing_df)

print("\nDrop remaining missing rows:")
print(missing_df.dropna())
```

```python
print("\n--- DUPLICATES ---")

duplicate_df = pd.concat([df, df.iloc[[0]]], ignore_index=True)

print("Duplicated rows:")
print(duplicate_df[duplicate_df.duplicated()])

print("\nAfter removing duplicates:")
duplicate_df = duplicate_df.drop_duplicates()

print(duplicate_df)
```

```python
print("\n--- DATA TYPE CONVERSION ---")

type_df = df.copy()

type_df["salary"] = type_df["salary"].astype(float)
type_df["age"] = type_df["age"].astype(int)
type_df["joining_date"] = pd.to_datetime(type_df["joining_date"])

print(type_df.dtypes)
```

```python
print("\n--- STRING OPERATIONS ---")

string_df = df.copy()

print("Uppercase names:")
print(string_df["name"].str.upper())

print("\nLowercase cities:")
print(string_df["city"].str.lower())

print("\nNames with A:")
print(string_df[string_df["name"].str.startswith("A")])

print("\nCities containing Pune:")
print(string_df[string_df["city"].str.contains("Pune")])

print("\nName lengths:")
print(string_df["name"].str.len())

print("\nReplace Pune with PUNE:")
print(string_df["city"].str.replace("Pune", "PUNE"))
```

```python
print("\n--- DATE AND TIME ---")

date_df = df.copy()

date_df["joining_date"] = pd.to_datetime(date_df["joining_date"])

date_df["joining_year"] = date_df["joining_date"].dt.year
date_df["joining_month"] = date_df["joining_date"].dt.month
date_df["joining_day"] = date_df["joining_date"].dt.day
date_df["day_of_week"] = date_df["joining_date"].dt.dayofweek
date_df["day_name"] = date_df["joining_date"].dt.day_name()

print(date_df[
    [
        "name",
        "joining_date",
        "joining_year",
        "joining_month",
        "joining_day",
        "day_of_week",
        "day_name"
    ]
])

print("\nDifference between joining dates:")
print(date_df["joining_date"].diff())

print("\nPrevious joining date:")
print(date_df["joining_date"].shift())
```

```python
print("\n--- SORTING ---")

print("Salary ascending:")
print(df.sort_values("salary"))

print("\nSalary descending:")
print(df.sort_values("salary", ascending=False))

print("\nSort by department and salary:")
print(df.sort_values(["department", "salary"]))

print("\nSort index:")
print(df.sort_index())
```

```python
print("\n--- GROUPBY ---")

print("Average salary by department:")
print(df.groupby("department")["salary"].mean())

print("\nTotal salary by department:")
print(df.groupby("department")["salary"].sum())

print("\nEmployee count by department:")
print(df.groupby("department")["employee_id"].count())

print("\nMaximum salary by department:")
print(df.groupby("department")["salary"].max())

print("\nMinimum salary by department:")
print(df.groupby("department")["salary"].min())

print("\nGroup by department and city:")
print(
    df.groupby(["department", "city"])["salary"].mean()
)
```

```python
print("\n--- AGGREGATION ---")

result = df.groupby("department")["salary"].agg(
    ["mean", "min", "max", "sum", "count", "median", "std"]
)

print(result)
```

```python
print("\n--- TRANSFORM ---")

transform_df = df.copy()

transform_df["department_avg_salary"] = (
    transform_df
    .groupby("department")["salary"]
    .transform("mean")
)

print(transform_df[
    [
        "name",
        "department",
        "salary",
        "department_avg_salary"
    ]
])

print("\nEmployees earning more than department average:")

print(
    transform_df[
        transform_df["salary"] >
        transform_df["department_avg_salary"]
    ]
)
```

```python
print("\n--- APPLY / MAP / LAMBDA ---")

apply_df = df.copy()

apply_df["salary_category"] = apply_df["salary"].apply(
    lambda x:
        "High" if x >= 80000
        else "Medium" if x >= 50000
        else "Low"
)

print(apply_df[
    ["name", "salary", "salary_category"]
])

# MAP
department_mapping = {
    "IT": "Technology",
    "HR": "Human Resources",
    "Finance": "Financial Department",
    "Sales": "Business Development"
}

apply_df["department_name"] = apply_df["department"].map(
    department_mapping
)

print("\nMapped departments:")
print(apply_df[
    ["department", "department_name"]
])
```

```python
print("\n--- CONCATENATION ---")

employees_jan = pd.DataFrame({
    "employee_id": [101, 102],
    "name": ["Aakanksha", "Rahul"],
    "salary": [55000, 65000]
})

employees_feb = pd.DataFrame({
    "employee_id": [103, 104],
    "name": ["Priya", "Sneha"],
    "salary": [70000, 75000]
})

print("Row-wise concatenation:")
print(
    pd.concat(
        [employees_jan, employees_feb],
        ignore_index=True
    )
)

print("\nColumn-wise concatenation:")

extra_data = pd.DataFrame({
    "city": ["Pune", "Mumbai"]
})

print(
    pd.concat(
        [employees_jan, extra_data],
        axis=1
    )
)
```

```python
print("\n---BASIC DATA TRANSFORMATION ---")

transform_df = df.copy()

# Rename column
transform_df = transform_df.rename(
    columns={"salary": "monthly_salary"}
)

print("Renamed column:")
print(transform_df.columns)

# Create calculated column
transform_df["annual_salary"] = (
    transform_df["monthly_salary"] * 12
)

transform_df["salary_after_bonus"] = (
    transform_df["monthly_salary"] * 1.10
)

print("\nCalculated columns:")
print(
    transform_df[
        [
            "name",
            "monthly_salary",
            "annual_salary",
            "salary_after_bonus"
        ]
    ]
)

# Drop column
transform_df = transform_df.drop(
    columns=["salary_after_bonus"]
)

print("\nAfter dropping column:")
print(transform_df.columns)

# Replace values
transform_df["city"] = transform_df["city"].replace(
    "Pune",
    "PUNE"
)

print("\nAfter replacing Pune:")
print(transform_df["city"])
```

```python
print("\n--- PIVOT TABLE ---")

pivot_result = pd.pivot_table(
    df,
    values="salary",
    index="department",
    columns="city",
    aggfunc="mean"
)

print("Average salary by department and city:")
print(pivot_result)

count_result = pd.pivot_table(
    df,
    values="employee_id",
    index="department",
    columns="city",
    aggfunc="count",
    fill_value=0
)

print("\nEmployee count by department and city:")
print(count_result)
```
