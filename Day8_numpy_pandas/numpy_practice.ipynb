```python
# impoerting libraries
import numpy as np
```

```python
print("\n---ARRAY CREATION ---")

arr = np.array([10, 20, 30, 40, 50])

print("Array:", arr)

print("Zeros:", np.zeros(5))
print("Ones:", np.ones(5))
print("Full:", np.full(5, 7))
print("Arange:", np.arange(1, 11))
print("Linspace:", np.linspace(1, 10, 5))
```

```python

print("\n---ARRAY PROPERTIES ---")

arr = np.array([[10, 20, 30],
                [40, 50, 60]])

print("Array:\n", arr)
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data type:", arr.dtype)
print("Item size:", arr.itemsize)
```

```python
print("\n---AXIS ---")

arr = np.array([[10, 20, 30],
                [40, 50, 60],
                [70, 80, 90]])

print("Array:\n", arr)

print("Column-wise sum (axis=0):", np.sum(arr, axis=0))
print("Row-wise sum (axis=1):", np.sum(arr, axis=1))

print("Column-wise mean:", np.mean(arr, axis=0))
print("Row-wise mean:", np.mean(arr, axis=1))
```

```python
print("\n---INDEXING ---")

arr = np.array([10, 20, 30, 40, 50])

print("First element:", arr[0])
print("Last element:", arr[-1])
print("Third element:", arr[2])

arr2 = np.array([[10, 20, 30],
                 [40, 50, 60]])

print("First row:", arr2[0])
print("Second column:", arr2[:, 1])
print("Element at row 2, column 3:", arr2[1, 2])
```

```python
print("\n--- SLICING ---")

arr = np.arange(10, 51)

print("Array:", arr)
print("First 3:", arr[:3])
print("Last 3:", arr[-3:])
print("Every second element:", arr[::2])
print("Index 2 to 6:", arr[2:7])
```

```python
print("\n--- BOOLEAN INDEXING ---")

marks = np.array([45, 78, 92, 35, 67, 88, 56, 95])

print("Marks:", marks)

print("Marks > 60:", marks[marks > 60])
print("Marks < 50:", marks[marks < 50])
print("Marks between 50 and 80:",
      marks[(marks >= 50) & (marks <= 80)])
```

```python
print("\n--- DATA TYPE CONVERSION ---")

arr = np.array([10, 20, 30, 40])

print("Original:", arr)
print("Original type:", arr.dtype)

float_arr = arr.astype(float)

print("Float array:", float_arr)
print("New type:", float_arr.dtype)

string_arr = arr.astype(str)

print("String array:", string_arr)
print("New type:", string_arr.dtype)
```

```python
print("\n--- MATHEMATICAL OPERATIONS ---")

a = np.array([10, 20, 30, 40])
b = np.array([2, 4, 5, 8])

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)

print("np.add():", np.add(a, b))
print("np.subtract():", np.subtract(a, b))
print("np.multiply():", np.multiply(a, b))
print("np.divide():", np.divide(a, b))

print("Square root:", np.sqrt(a))
```

```python

print("\n--- STATISTICAL FUNCTIONS ---")

marks = np.array([45, 78, 92, 35, 67, 88, 56, 95])

print("Marks:", marks)
print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard deviation:", np.std(marks))
print("Variance:", np.var(marks))
```

```python
print("\n---  NaN AND SPECIAL VALUES ---")

arr = np.array([10, 20, np.nan, 40, np.inf, 60])

print("Array:", arr)

print("NaN positions:", np.isnan(arr))
print("Infinite positions:", np.isinf(arr))

valid_values = arr[~np.isnan(arr) & ~np.isinf(arr)]

print("Valid values:", valid_values)
print("Mean of valid values:", np.mean(valid_values))
```

```python

print("\n--- RESHAPING ---")

arr = np.arange(1, 13)

print("Original:", arr)

print("2 x 6:\n", arr.reshape(2, 6))
print("3 x 4:\n", arr.reshape(3, 4))
print("4 x 3:\n", arr.reshape(4, 3))

matrix = arr.reshape(3, 4)

print("Flatten:", matrix.flatten())
print("Ravel:", matrix.ravel())

print("Transpose:\n", matrix.transpose())
```

```python
print("\n--- COMBINING ARRAYS ---")

a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Concatenate:", np.concatenate((a, b)))

print("Vertical stack:")
print(np.vstack((a, b)))

print("Horizontal stack:")
print(np.hstack((a, b)))

print("Stack:")
print(np.stack((a, b)))
```

```python
print("\n--- SPLITTING ARRAYS ---")

arr = np.arange(1, 13)

print("Array:", arr)

print("2 parts:")
print(np.split(arr, 2))

print("3 parts:")
print(np.split(arr, 3))

print("Array split into 5 parts:")
print(np.array_split(arr, 5))
```

```python
print("\n--- SORTING AND SEARCHING ---")

arr = np.array([50, 20, 80, 10, 40, 80, 30])

print("Original:", arr)
print("Sorted:", np.sort(arr))
print("Argsort:", np.argsort(arr))
print("Maximum position:", np.argmax(arr))
print("Minimum position:", np.argmin(arr))
print("Unique values:", np.unique(arr))
```

```python
print("\n--- WHERE ---")

marks = np.array([45, 78, 92, 35, 67, 88, 40])

result = np.where(marks >= 50, "Pass", "Fail")

print("Marks:", marks)
print("Result:", result)
```

```python
print("\n--- BROADCASTING ---")

arr = np.array([10, 20, 30, 40])

print("Original:", arr)
print("After adding 5:", arr + 5)

matrix = np.array([[10, 20, 30],
                   [40, 50, 60]])

values = np.array([1, 2, 3])

print("Matrix:\n", matrix)
print("Values:", values)
print("Broadcasting result:\n", matrix + values)
```

```python
print("\n--- RANDOM NUMBERS ---")

np.random.seed(42)

print("Random integers:")
print(np.random.randint(1, 101, 10))

print("Random 3 x 3 array:")
print(np.random.rand(3, 3))

print("Random normal values:")
print(np.random.randn(5))
```
